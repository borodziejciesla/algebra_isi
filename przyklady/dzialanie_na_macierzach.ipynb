# Działania na macierzach w SymPy i NumPy

Pokażemy podstawowe działania na macierzach i rozwiązywanie układu równań w dwóch bibliotekach. SymPy pozwala zachować wyniki dokładne, a NumPy wykonuje obliczenia numeryczne na tablicach.

```python
import numpy as np
import sympy as sp

A_sym = sp.Matrix([
    [1, 2],
    [3, 4],
])

B_sym = sp.Matrix([
    [5, 6],
    [7, 8],
])

A_np = np.array([
    [1, 2],
    [3, 4],
])
B_np = np.array([
    [5, 6],
    [7, 8],
])

print("Macierz A w SymPy:")
sp.pprint(A_sym)
print("Wymiary A w SymPy:", A_sym.shape)
print("Macierz A w NumPy:")
print(A_np)
print("Wymiary A w NumPy:", A_np.shape)
```

## Dodawanie, odejmowanie i mnożenie przez skalar

Dodawanie i odejmowanie wykonują się element po elemencie. Macierze muszą mieć takie same wymiary. Mnożenie macierzy przez liczbę również skaluje każdy jej element.

```python
print("A + B w SymPy =")
sp.pprint(A_sym + B_sym)
print("A + B w NumPy =")
print(A_np + B_np)

print("A - B w SymPy =")
sp.pprint(A_sym - B_sym)
print("A - B w NumPy =")
print(A_np - B_np)

print("2 * A w SymPy =")
sp.pprint(2 * A_sym)
print("2 * A w NumPy =")
print(2 * A_np)
```

## Mnożenie elementowe a mnożenie macierzy

W SymPy operator `*` dla dwóch macierzy oznacza mnożenie macierzy, a iloczyn elementowy oblicza metoda `multiply_elementwise`. W NumPy operator `*` mnoży elementy na tych samych pozycjach, a `@` oznacza mnożenie macierzy. Dla $A$ o wymiarach $m\times n$ i $B$ o wymiarach $n\times k$ wynik $AB$ ma wymiary $m\times k$.

```python
print("Mnożenie elementowe w SymPy:")
sp.pprint(A_sym.multiply_elementwise(B_sym))
print("Mnożenie elementowe w NumPy (A_np * B_np):")
print(A_np * B_np)

print("Mnożenie macierzy w SymPy (A_sym * B_sym):")
sp.pprint(A_sym * B_sym)
print("Mnożenie macierzy w NumPy (A_np @ B_np):")
print(A_np @ B_np)
```

## Transpozycja i wyznacznik

Atrybut `.T` zamienia wiersze z kolumnami w obu bibliotekach. Wyznacznik w SymPy oblicza metoda `.det()` i zwraca wynik dokładny; w NumPy służy do tego `np.linalg.det`, która zwraca przybliżenie zmiennoprzecinkowe.

```python
print("Transpozycja A w SymPy:")
sp.pprint(A_sym.T)
print("Transpozycja A w NumPy:")
print(A_np.T)

print("Wyznacznik A w SymPy:", A_sym.det())
print("Wyznacznik A w NumPy:", np.linalg.det(A_np))
```

## Rozwiązywanie układu równań

Rozważmy układ

$$
\begin{cases}
3x + y = 9,\\
x + 2y = 8.
\end{cases}
$$

Zapisujemy go jako $Cx=b$. SymPy rozwiązuje układ metodą `LUsolve`, zwracając dokładny wynik. NumPy używa `np.linalg.solve` i oblicza rozwiązanie numerycznie. Macierz `C` jest kwadratowa i odwracalna.

```python
C_sym = sp.Matrix([
    [3, 1],
    [1, 2],
])
b_sym = sp.Matrix([9, 8])

x_sym = C_sym.LUsolve(b_sym)

print("Dokładne rozwiązanie w SymPy [x, y]:")
sp.pprint(x_sym)
print("Sprawdzenie w SymPy, C * x =")
sp.pprint(C_sym * x_sym)

C_np = np.array([
    [3, 1],
    [1, 2],
])
b_np = np.array([9, 8])
x_np = np.linalg.solve(C_np, b_np)

print("Numeryczne rozwiązanie w NumPy [x, y]:", x_np)
print("Sprawdzenie w NumPy, C @ x =", C_np @ x_np)
```

Obie biblioteki dają rozwiązanie $x=2$, $y=3$. SymPy zachowuje dokładne wartości, a NumPy zwraca przybliżenia numeryczne. W obu przypadkach układ rozwiązujemy bez jawnego obliczania macierzy odwrotnej.
